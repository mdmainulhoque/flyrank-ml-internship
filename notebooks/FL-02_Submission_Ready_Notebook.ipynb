````python
## Version 5

### Prompt
```text
1.Read dataset
2.Clean data
3.Rank keywords
4.Assign scores
5.Recommend Top 20
6.Explain results
```

### ChatGPT Output
import pandas as pd

# Read dataset
df = pd.read_csv("content_refresh_anonymized(3).csv")

# Keep required columns
cols = [
    "content_id",
    "search_volume",
    "competition",
    "competition_level",
    "cpc",
    "main_intent"
]

df = df[cols]

# Remove invalid rows
df = df.dropna()
df = df[
    (df["search_volume"] >= 0) &
    (df["competition"] >= 0) &
    (df["cpc"] >= 0)
]

# Normalize metrics
df["sv_score"] = df["search_volume"] / df["search_volume"].max()
df["cpc_score"] = df["cpc"] / max(df["cpc"].max(), 1)

# Lower competition = better
df["competition_score"] = 1 - df["competition"]

intent_bonus = {
    "transactional": 1.0,
    "commercial": 0.9,
    "informational": 0.6,
    "navigational": 0.4
}

df["intent_score"] = (
    df["main_intent"]
      .str.lower()
      .map(intent_bonus)
      .fillna(0.5)
)

# Opportunity Score (1–10)
score = (
    0.40 * df["sv_score"] +
    0.25 * df["competition_score"] +
    0.20 * df["cpc_score"] +
    0.15 * df["intent_score"]
)

df["Opportunity Score"] = (score * 9 + 1).round(1)

# Rank
df = df.sort_values(
    "Opportunity Score",
    ascending=False
)

top20 = df.head(20).copy()

def recommendation(row):
    if row["main_intent"].lower() == "transactional":
        return "Prioritize for conversion-focused landing pages."
    elif row["main_intent"].lower() == "commercial":
        return "Target with product comparisons and buying guides."
    else:
        return "Create comprehensive informational content."

def reason(row):
    return (
        f"Search Volume={row.search_volume}, "
        f"Competition={row.competition_level}, "
        f"CPC=${row.cpc:.2f}, "
        f"Intent={row.main_intent}."
    )

top20["Recommendation"] = top20.apply(recommendation, axis=1)
top20["Reason"] = top20.apply(reason, axis=1)

print(
    top20[
        [
            "content_id",
            "search_volume",
            "competition_level",
            "cpc",
            "main_intent",
            "Opportunity Score",
            "Recommendation",
            "Reason"
        ]
    ].to_markdown(index=False)
)

### Iteration Analysis
Task decomposition improves reasoning.
````
